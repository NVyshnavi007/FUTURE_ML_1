```python
ts = monthly_sales.reset_index()
ts.columns = ['Month_Period', 'Sales']
ts['Month_Period'] = ts['Month_Period'].dt.to_timestamp()
ts = ts.sort_values('Month_Period').reset_index(drop=True)

# continuous time index -> captures overall trend
ts['t'] = range(len(ts))

# calendar features
ts['Month'] = ts['Month_Period'].dt.month
ts['Quarter'] = ts['Month_Period'].dt.quarter

# cyclical encoding so December and January are 'close' to the model, not 11 apart
ts['Month_sin'] = np.sin(2 * np.pi * ts['Month'] / 12)
ts['Month_cos'] = np.cos(2 * np.pi * ts['Month'] / 12)

ts.head()
```

```python
split_idx = int(len(ts) * 0.8)
train = ts.iloc[:split_idx]
test = ts.iloc[split_idx:]

print(f"Train months: {len(train)}  |  Test months: {len(test)}")
print(f"Train range: {train['Month_Period'].min().date()} to {train['Month_Period'].max().date()}")
print(f"Test range:  {test['Month_Period'].min().date()} to {test['Month_Period'].max().date()}")
```

```python
from sklearn.linear_model import LinearRegression

features = ['t', 'Month_sin', 'Month_cos', 'Quarter']

X_train, y_train = train[features], train['Sales']
X_test, y_test = test[features], test['Sales']

model = LinearRegression()
model.fit(X_train, y_train)
model.score(X_test,y_test)

preds = model.predict(X_test)
preds
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(y_test, preds)
rmse = mean_squared_error(y_test, preds) ** 0.5
mape = (abs((y_test - preds) / y_test)).mean() * 100

print(f"MAE:  ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"MAPE: {mape:.1f}%")
```

```python
future_periods = 6
last_t = ts['t'].iloc[-1]
last_date = ts['Month_Period'].iloc[-1]

future_dates = pd.date_range(last_date + pd.DateOffset(months=1), periods=future_periods, freq='MS')

future = pd.DataFrame({'Month_Period': future_dates})
future['t'] = range(last_t + 1, last_t + 1 + future_periods)
future['Month'] = future['Month_Period'].dt.month
future['Quarter'] = future['Month_Period'].dt.quarter
future['Month_sin'] = np.sin(2 * np.pi * future['Month'] / 12)
future['Month_cos'] = np.cos(2 * np.pi * future['Month'] / 12)

future['Forecast'] = model.predict(future[features])
future[['Month_Period', 'Forecast']]
```

```python
plt.figure(figsize=(12, 5))
plt.plot(ts['Month_Period'], ts['Sales'], label='Actual Sales', marker='o')
plt.plot(future['Month_Period'], future['Forecast'], label='Forecasted Sales',
         marker='o', linestyle='--', color='red')

plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.title('Monthly Sales — Actual vs. 6-Month Forecast')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
## Step 12: Business explanation
#What the forecast shows:
#This model looks at past sales and tries to guess sales for the next 6 months. 
#It found that sales usually go up near the end of the year (around Q4), 
#probably because of holiday shopping. 
#(Put your real numbers here once you run the code — for example: 
#the next 3 months are expected to be about X% higher than last year.)

#How a business can use this:
#- Stock/Inventory:Order more products before the months where sales are expected to go up, so the store doesn't run out.
#- Money planning: In months where sales are expected to be low, the business can spend less on non-important things.
#- Staff:Hire extra workers a little before the busy season starts, not after it already started.

#How accurate is this model:
#On average, the model's guess was off by about [19.3]% for random forest and [20.5%] for linear regression. 
#This is good enough to help with planning, but it's not 100% exact
#it should be used as a helpful guide, not a promise.
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=200, max_depth=4, random_state=42)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
```

```python
rf_mae = mean_absolute_error(y_test, rf_preds)
rf_rmse = mean_squared_error(y_test, rf_preds) ** 0.5
rf_mape = (abs((y_test - rf_preds) / y_test)).mean() * 100

print("Linear Regression:")
print(f"  MAE:  ${mae:,.2f}")
print(f"  RMSE: ${rmse:,.2f}")
print(f"  MAPE: {mape:.1f}%")

print("\nRandom Forest:")
print(f"  MAE:  ${rf_mae:,.2f}")
print(f"  RMSE: ${rf_rmse:,.2f}")
print(f"  MAPE: {rf_mape:.1f}%")
```

```python
final_model = rf_model  

future['Forecast'] = final_model.predict(future[features])
future[['Month_Period', 'Forecast']]
```
